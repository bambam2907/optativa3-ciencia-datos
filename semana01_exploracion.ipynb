## Dataset: Spotify Tracks Dataset

**Origen.** Kaggle, publicado por el usuario *maharshipandya*, con datos obtenidos de la Spotify
Web API: <https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset>

**Registros.** El archivo original tiene 114,000 canciones. Para trabajar más ágil tomé una
muestra aleatoria de **200 canciones** (semilla 42), guardada como `spotify_200.csv`. Son 200
filas × 20 columnas, con 93 géneros distintos y sin valores faltantes.

**Variables.** Datos de identificación: `track_id`, `artists`, `album_name`, `track_name`,
`track_genre` y `explicit`. Medidas numéricas: `popularity` (0–100), `duration_ms`, `loudness`
(dB), `tempo` (BPM), `key`, `mode` y `time_signature`. Y las *audio features* que calcula Spotify
en escala de 0 a 1: `danceability`, `energy`, `speechiness`, `acousticness`, `instrumentalness`,
`liveness` y `valence`.

**Por qué lo elegí.** Escucho música todos los días y me llamó la atención que Spotify guarde un
perfil numérico de cada canción que como usuario nunca veo. Me pareció interesante poder medir
algo tan subjetivo como si una canción es alegre o bailable.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carga del dataset elegido
df = pd.read_csv("spotify_200.csv")
```

```python
# Exploracion inicial
print("Dimensiones (filas, columnas):", df.shape)

print("\nPrimeras filas:")
print(df.head())

print("\nTipos de datos:")
print(df.dtypes)

print("\nResumen estadistico:")
print(df.describe())
```

```python
# Primer grafico: distribucion de una columna numerica
columna = df.select_dtypes("number").columns[0]

df[columna].hist(bins=20)
plt.title(f"Distribucion de {columna}")
plt.xlabel(columna); plt.ylabel("Frecuencia")
plt.tight_layout()
plt.savefig("distribucion.png")
plt.show()
```
